````python
import re

clean_report = staff_report_text

# Join words split across PDF line breaks
clean_report = re.sub(r'(\w)-\s*\n\s*(\w)', r'\1\2', clean_report)

# Remove excessive spaces and blank lines
clean_report = re.sub(r'[ \t]+', ' ', clean_report)
clean_report = re.sub(r'\n{3,}', '\n\n', clean_report)

clean_report = clean_report.strip()

print("Original characters:", len(staff_report_text))
print("Cleaned characters:", len(clean_report))
print("\nCleaned preview:\n")
print(clean_report[:1500])
````

````python
import tiktoken

# Load tokenizer
encoding = tiktoken.get_encoding("cl100k_base")

# Convert report into tokens
tokens = encoding.encode(clean_report)

# Chunk settings
chunk_size = 400
overlap = 50
step = chunk_size - overlap

chunks = []

for start in range(0, len(tokens), step):
    end = start + chunk_size

    chunk_tokens = tokens[start:end]
    chunk_text = encoding.decode(chunk_tokens)

    if chunk_text.strip():
        chunks.append(chunk_text)

print("Total tokens:", len(tokens))
print("Total chunks:", len(chunks))
print("\nFirst chunk:\n")
print(chunks[0])
````

````python
chunk_lengths = [len(encoding.encode(chunk)) for chunk in chunks]

print("Smallest chunk:", min(chunk_lengths))
print("Largest chunk:", max(chunk_lengths))
print("Average chunk size:", round(sum(chunk_lengths) / len(chunk_lengths), 2))

print("\nFirst 5 chunk sizes:")
print(chunk_lengths[:5])
````

````python
from sentence_transformers import SentenceTransformer
import numpy as np

# Load a lightweight embedding model
embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

# Convert all 171 chunks into embeddings
embeddings = embedding_model.encode(
    chunks,
    convert_to_numpy=True,
    show_progress_bar=True
)

print("Number of chunks:", len(chunks))
print("Embedding shape:", embeddings.shape)
````

````python
import faiss

# Convert embeddings to float32
embeddings = embeddings.astype("float32")

# Normalize embeddings
faiss.normalize_L2(embeddings)

# Create a FAISS index
dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

# Add embeddings to the index
index.add(embeddings)

print("FAISS index created successfully.")
print("Total indexed chunks:", index.ntotal)
````

````python
def retrieve_context(query, top_k=3):

    # Convert the question into an embedding
    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True
    ).astype("float32")

    # Normalize the question embedding
    faiss.normalize_L2(query_embedding)

    # Search FAISS for the closest chunks
    distances, indices = index.search(query_embedding, top_k)

    results = []

    for rank, chunk_index in enumerate(indices[0]):

        results.append({
            "rank": rank + 1,
            "distance": float(distances[0][rank]),
            "chunk_index": int(chunk_index),
            "text": chunks[chunk_index]
        })

    return results
````

````python
test_query = "What are the main risks to UK financial stability?"

retrieved_results = retrieve_context(test_query, top_k=3)

for result in retrieved_results:
    print(f"\n--- Retrieved Chunk {result['chunk_index']} ---")
    print("Distance:", round(result["distance"], 4))
    print(result["text"][:1000])
````

````python
def rag_answer(query, top_k=3):

    # Retrieve relevant passages
    results = retrieve_context(query, top_k=top_k)

    # Combine retrieved passages with their chunk references
    context = ""

    for result in results:
        context += f"""
        SOURCE CHUNK {result['chunk_index']}:
        {result['text']}

        """

    # Create a grounded answer prompt
    rag_prompt = f"""
    You are an economic research assistant.

    Answer the question using ONLY the supplied IMF report passages.

    RULES:
    - Do not introduce information from outside the passages.
    - Do not invent statistics or claims.
    - If the passages do not contain enough information,
      clearly state that.
    - Keep the answer concise and factual.
    - Identify the source chunk numbers supporting your answer.

    QUESTION:
    {query}

    RETRIEVED IMF REPORT PASSAGES:
    {context}
    """

    # Generate the answer
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=rag_prompt
    )

    return {
        "question": query,
        "answer": response.text,
        "sources": [
            {
                "chunk_index": r["chunk_index"],
                "distance": r["distance"]
            }
            for r in results
        ]
    }
````

````python
# Original research question
original_query = "What are the main risks to UK financial stability?"

# Three alternative search queries
multi_queries = [
    original_query,
    "What vulnerabilities exist in UK financial markets, including the gilt market?",
    "What risks do non-bank financial institutions pose to the UK banking system?",
    "How could global economic uncertainty and geopolitical shocks affect UK financial stability?"
]

print("Original question:")
print(original_query)

print("\nMulti-query variations:")
for i, query in enumerate(multi_queries, 1):
    print(f"{i}. {query}")
````

````python
def multi_query_retrieval(queries, top_k=3):

    combined_results = {}

    for query in queries:

        results = retrieve_context(query, top_k=top_k)

        for result in results:

            chunk_id = result["chunk_index"]

            # Keep the best (lowest) distance for duplicate chunks
            if (
                chunk_id not in combined_results
                or result["distance"] < combined_results[chunk_id]["distance"]
            ):
                combined_results[chunk_id] = result

    # Sort by retrieval distance
    final_results = sorted(
        combined_results.values(),
        key=lambda x: x["distance"]
    )

    return final_results


multi_results = multi_query_retrieval(
    multi_queries,
    top_k=3
)

print("Unique chunks retrieved:", len(multi_results))

for result in multi_results:
    print(
        f"\nChunk: {result['chunk_index']}"
        f"\nDistance: {result['distance']:.4f}"
        f"\nText: {result['text'][:500]}..."
    )
````

````python
def multi_query_rag_answer(query, queries, top_k=3):

    results = multi_query_retrieval(
        queries,
        top_k=top_k
    )

    context = ""

    for result in results:
        context += f"""
SOURCE CHUNK {result['chunk_index']}:
{result['text']}

"""

    rag_prompt = f"""
You are an economic research assistant.

Answer the question using ONLY the supplied IMF report passages.

RULES:
- Do not introduce outside information.
- Do not invent statistics or claims.
- Clearly state when evidence is insufficient.
- Keep the answer factual and organized.
- Identify the source chunk numbers supporting each claim.

QUESTION:
{query}

RETRIEVED IMF REPORT PASSAGES:
{context}
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=rag_prompt
    )

    return {
        "question": query,
        "answer": response.text,
        "sources": [
            {
                "chunk_index": r["chunk_index"],
                "distance": r["distance"]
            }
            for r in results
        ]
    }


multi_rag_result = multi_query_rag_answer(
    original_query,
    multi_queries,
    top_k=3
)

print("MULTI-QUERY RAG ANSWER:\n")
print(multi_rag_result["answer"])

print("\nRETRIEVED SOURCES:")
for source in multi_rag_result["sources"]:
    print(source)
````
